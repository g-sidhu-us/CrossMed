# CrossMed 02 - Train models
Train ResNet-18 first, then optionally ViT-B/16 using the same manifests.

```python
from pathlib import Path
import pandas as pd
import torch
from crossmed.data import UltrasoundDataset
from crossmed.models import build_model
from crossmed.train import train_experiment
print('PyTorch:', torch.__version__, 'CUDA:', torch.cuda.is_available())
if torch.cuda.is_available(): print(torch.cuda.get_device_name(0))
```

## Inspect one transformed sample

```python
dataset = UltrasoundDataset('../data/manifests/busbra_train.csv', image_size=224, train=True)
sample = dataset[0]
print(sample['image'].shape, sample['label'].item(), sample['sample_id'])
```

## Check model output before a long run

```python
model = build_model('resnet18', pretrained=False)
with torch.no_grad():
    output = model(sample['image'].unsqueeze(0))
print('Output shape:', output.shape)
del model
```

## Train ResNet-18 baseline

```python
resnet_run = train_experiment('../configs/baseline.yaml')
pd.DataFrame(resnet_run['history']).tail()
```
