# CrossMed 04 - Experiment matrix
Run one, several, or all planned architectures without duplicating training code. Completed stages are skipped by default.

```python
from pathlib import Path
import pandas as pd
from crossmed.experiments import load_experiment_matrix, run_experiment_matrix, aggregate_results
PROJECT_ROOT = Path('..').resolve() if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
MATRIX = PROJECT_ROOT / 'configs/experiment_matrix.yaml'
```

## View available experiments

```python
experiments = load_experiment_matrix(MATRIX)
pd.DataFrame({'experiment': experiments.keys(), 'config': experiments.values()})
```

## Select runs
Start with one experiment. Add more names after confirming that data loading and GPU execution work.

```python
SELECTED = ['resnet18_finetuned']
# Full suite:
# SELECTED = list(experiments)
```

## Train selected experiments

```python
run_experiment_matrix(MATRIX, selected=SELECTED, stage='train', skip_completed=True)
```

## Evaluate selected experiments

```python
comparison = run_experiment_matrix(MATRIX, selected=SELECTED, stage='evaluate', skip_completed=True)
comparison
```

## Reload available results without training
This cell is safe to use later for reports and presentations.

```python
comparison = aggregate_results(MATRIX)
comparison
```

```python
if not comparison.empty:
    display(comparison.pivot(index='experiment', columns='dataset', values='roc_auc'))
```
